Game Walkthrough Text

This is the game guide text that our system will use to answer questions.

```python
# Sample Game Walkthrough Text (The Context)
game_guide = """
=== THE SHADOW REALM: COMPLETE WALKTHROUGH ===

CHAPTER 1: THE BEGINNING
Your journey begins in the Village of Elmsworth. Your main objective is to find the Ancient Amulet hidden deep within the Shadow Realm. 
To start, speak with the Elder in the village square. He will give you the Rusty Sword and 50 gold coins.

CHAPTER 2: THE HIDDEN KEY
The hidden key to the Ancient Temple is located behind the waterfall in the Misty Forest. To reach it, you must first defeat the 
Forest Guardian, a level 15 wolf boss. Once defeated, swim through the waterfall and you'll find the Golden Key embedded in a stone 
pedestal. Click on the pedestal to collect the key. This key opens the main gate to the Ancient Temple.

CHAPTER 3: THE ANCIENT TEMPLE PUZZLE
Inside the temple, you'll encounter the Door Puzzle in the Hall of Mirrors. To solve this puzzle, you need to arrange the colored 
crystals in the following order from left to right: Red, Blue, Green, Yellow. Each crystal must match the color of the torch directly 
above it. Once correctly aligned, the door will automatically open, revealing the Treasure Chamber.

CHAPTER 4: ESSENTIAL ITEMS
Throughout your journey, you'll need these critical items:
- Health Potion: Found in wooden crates scattered throughout dungeons. Restores 50 HP.
- Mana Crystal: Dropped by Mage enemies. Restores 30 MP.
- Warp Stone: Purchase from the Merchant in Elmsworth for 200 gold. Allows instant teleportation to any visited town.
- Dragon Scale Armor: Crafted by the Blacksmith using 3 Dragon Scales (dropped by the Dragon boss in Mount Inferno) and 500 gold.

CHAPTER 5: THE FINAL BOSS
The final boss is Lord Malakar, the Shadow King. He resides in the Throne Room at the top of the Shadow Tower. Lord Malakar has 
two phases: In Phase 1, he uses dark magic attacks. Stay at range and dodge his shadow bolts. In Phase 2 (below 50% HP), he summons 
shadow minions. Focus on eliminating the minions first before attacking Malakar. His weakness is fire magic, so equip fire-enchanted 
weapons or use Fire Spells for maximum damage.

CHAPTER 6: SECRET QUESTS
There are three hidden quests:
1. The Lost Artifact: Talk to the Librarian in the Royal Library. She asks you to find the Crystal Orb hidden in the Catacombs beneath 
   the castle. Reward: 1000 XP and the Staff of Wisdom.
2. The Cursed Sword: A mysterious merchant in the Black Market sells a cursed sword for 1000 gold. To lift the curse, bring it to 
   the Priest in the Cathedral. Reward: The sword transforms into the Legendary Blade of Light.
3. The Dragon's Lair: Accessible only after completing Chapter 4. Defeat the Ancient Dragon in Mount Inferno to obtain the Dragon 
   Heart, which can be used to craft the most powerful armor in the game.

TIPS AND TRICKS:
- Always save your game before entering boss battles.
- Stock up on at least 10 Health Potions before facing Lord Malakar.
- The best grinding spot for experience is the Goblin Camp in the Eastern Plains.
- To unlock the secret ending, you must complete all three hidden quests before defeating the final boss.
"""

print("Game guide loaded.")
print(f"Total words: {len(game_guide.split())}")
```

Load the Question Answering Model

Using DistilBERT model trained on SQuAD dataset for answering questions.

```python
# Initialize the Question Answering pipeline
qa_assistant = pipeline(
    "question-answering",
    model="distilbert-base-cased-distilled-squad",
    device=device
)
```

Testing the System

Asking different types of questions to test how well the system works.

```python
question_1 = "Where is the hidden key?"
result_1 = qa_assistant(question=question_1, context=game_guide)

print(f"Question: {question_1}")
print(f"Answer: {result_1['answer']}")
print(f"Confidence: {result_1['score']:.2f}")
print()
```

Summary of All Results

```python
# Collect all results
all_results = [
    ("Where is the hidden key?", result_1),
    ("What is the main objective?", result_2),
    ("How do I solve the door puzzle?", result_3),
    ("Who is the final boss?", result_4),
    ("What is Lord Malakar's weakness?", result_5),
    ("How do I get the Dragon Scale Armor?", result_6),
    ("How do I unlock the secret ending?", result_7),
    ("Where does the journey begin?", result_8)
]

print("\nAll Questions and Answers:")
print("-" * 80)

for idx, (question, result) in enumerate(all_results, 1):
    print(f"{idx}. {question}")
    print(f"   Answer: {result['answer']}")
    print(f"   Confidence: {result['score']:.2f}")
    print()
```

```python

# Statistics
scores = [r[1]['score'] for r in all_results]
avg_score = sum(scores) / len(scores)

print("\nStatistics:")
print(f"Total questions answered: {len(all_results)}")
print(f"Average confidence: {avg_score:.2f}")
print(f"Lowest confidence: {min(scores):.2f}")
print(f"Highest confidence: {max(scores):.2f}")
```

What I Learned

This project taught me how to use AI for question answering. The system worked pretty well overall.

Key Points:

1. The model answered all 8 questions correctly, though some answers were incomplete
2. Short, simple questions got higher confidence scores
3. Longer, complex questions had lower confidence but were still accurate
4. The model sometimes cuts off answers too early (for example, "Crafted by the Blacks" instead of "Blacksmith")
5. Confidence scores above 1.0 are just technical quirks - they mean the model is very confident

What Worked Well:
- Finding specific names and locations
- Answering "where", "who", and "what" questions
- Fast response time

What Could Be Better:
- Sometimes answers are too short and miss important details
- Can't combine information from multiple parts of the text
- Need to manually check if answers are complete

This would be useful for actual game players who need quick answers without reading through long guides.
