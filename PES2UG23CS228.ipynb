## The Models to Test
1.  **BERT** (`bert-base-uncased`): An **Encoder-only** model (designed for understanding, not generation).
2.  **RoBERTa** (`roberta-base`): An optimized **Encoder-only** model.
3.  **BART** (`facebook/bart-base`): An **Encoder-Decoder** model (designed for seq2seq tasks like translation/generation).

## Setup: Import Libraries

First, we'll import the necessary libraries and set up device configuration for GPU acceleration.

```python
from transformers import pipeline
import torch
import warnings
warnings.filterwarnings('ignore')
```

### Check GPU Availability

Let's check if CUDA is available for GPU acceleration.

```python
# Set up device for GPU acceleration
device = 0 if torch.cuda.is_available() else -1
if device == 0:
    print(f"✓ Using GPU: {torch.cuda.get_device_name(0)}")
else:
    print("✗ CUDA not available, using CPU")
```

## Experiment 1: Text Generation

**Task**: Try to generate text using the prompt: `"The future of Artificial Intelligence is"`

**Hypothesis**: Which models will fail? Why? (Hint: Can an Encoder *generate* new tokens easily?)

```python
prompt = "The future of Artificial Intelligence is"

print("="*60)
print("EXPERIMENT 1: TEXT GENERATION")
print("="*60)
```

### Test 1.1: BERT for Text Generation

```python
print("\n--- Testing BERT (bert-base-uncased) ---")
try:
    bert_gen = pipeline('text-generation', model='bert-base-uncased', device=device)
    result_bert = bert_gen(prompt, max_length=30)
    print(f"Result: {result_bert[0]['generated_text']}")
    print("Status: SUCCESS")
except Exception as e:
    print(f"Status: FAILURE")
    print(f"Error: {type(e).__name__}: {str(e)[:100]}")
```

```python
print("\n--- Testing BART (facebook/bart-base) ---")
try:
    bart_gen = pipeline('text-generation', model='facebook/bart-base', device=device)
    result_bart = bart_gen(prompt, max_length=30)
    print(f"Result: {result_bart[0]['generated_text']}")
    print("Status: SUCCESS")
except Exception as e:
    print(f"Status: FAILURE")
    print(f"Error: {type(e).__name__}: {str(e)[:100]}")
```

## Experiment 2: Masked Language Modeling (Fill-in-the-Blank)

**Task**: Predict the missing word in: `"The goal of Generative AI is to [MASK] new content."`

**Hypothesis**: This is how BERT/RoBERTa were trained. They should perform well.

```python
# Note: Different models use different mask tokens
masked_sentence_bert = "The goal of Generative AI is to [MASK] new content."
masked_sentence_roberta = "The goal of Generative AI is to <mask> new content."

print("="*60)
print("EXPERIMENT 2: MASKED LANGUAGE MODELING")
print("="*60)
```

### Test 2.1: BERT for Fill-Mask

```python
print("\n--- Testing BERT (bert-base-uncased) ---")
try:
    bert_mask = pipeline('fill-mask', model='bert-base-uncased', device=device)
    result_bert = bert_mask(masked_sentence_bert)
    print("Top 3 predictions:")
    for i, pred in enumerate(result_bert[:3], 1):
        print(f"  {i}. {pred['token_str']} (score: {pred['score']:.4f})")
    print("Status: SUCCESS")
except Exception as e:
    print(f"Status: FAILURE")
    print(f"Error: {type(e).__name__}: {str(e)[:100]}")
```

### Test 2.2: RoBERTa for Fill-Mask

```python
print("\n--- Testing RoBERTa (roberta-base) ---")
try:
    roberta_mask = pipeline('fill-mask', model='roberta-base', device=device)
    result_roberta = roberta_mask(masked_sentence_roberta)
    print("Top 3 predictions:")
    for i, pred in enumerate(result_roberta[:3], 1):
        print(f"  {i}. {pred['token_str']} (score: {pred['score']:.4f})")
    print("Status: SUCCESS")
except Exception as e:
    print(f"Status: FAILURE")
    print(f"Error: {type(e).__name__}: {str(e)[:100]}")
```

### Test 2.3: BART for Fill-Mask

```python
print("\n--- Testing BART (facebook/bart-base) ---")
try:
    bart_mask = pipeline('fill-mask', model='facebook/bart-base', device=device)
    result_bart = bart_mask(masked_sentence_roberta)  # BART uses <mask> like RoBERTa
    print("Top 3 predictions:")
    for i, pred in enumerate(result_bart[:3], 1):
        print(f"  {i}. {pred['token_str']} (score: {pred['score']:.4f})")
    print("Status: SUCCESS")
except Exception as e:
    print(f"Status: FAILURE")
    print(f"Error: {type(e).__name__}: {str(e)[:100]}")
```

## Experiment 3: Question Answering

**Task**: Answer the question `"What are the risks?"` based on the context: `"Generative AI poses significant risks such as hallucinations, bias, and deepfakes."`

**Note**: Using a "base" model (not fine-tuned for SQuAD) might yield random or poor results. Observe this behavior.

```python
context = "Generative AI poses significant risks such as hallucinations, bias, and deepfakes."
question = "What are the risks?"

print("="*60)
print("EXPERIMENT 3: QUESTION ANSWERING")
print("="*60)
print(f"Context: {context}")
print(f"Question: {question}")
```

### Test 3.1: BERT for Question Answering

```python
print("\n--- Testing BERT (bert-base-uncased) ---")
try:
    bert_qa = pipeline('question-answering', model='bert-base-uncased', device=device)
    result_bert = bert_qa(question=question, context=context)
    print(f"Answer: {result_bert['answer']}")
    print(f"Score: {result_bert['score']:.4f}")
    print("Status: SUCCESS")
except Exception as e:
    print(f"Status: FAILURE")
    print(f"Error: {type(e).__name__}: {str(e)[:100]}")
```

### Test 3.2: RoBERTa for Question Answering

```python
print("\n--- Testing RoBERTa (roberta-base) ---")
try:
    roberta_qa = pipeline('question-answering', model='roberta-base', device=device)
    result_roberta = roberta_qa(question=question, context=context)
    print(f"Answer: {result_roberta['answer']}")
    print(f"Score: {result_roberta['score']:.4f}")
    print("Status: SUCCESS")
except Exception as e:
    print(f"Status: FAILURE")
    print(f"Error: {type(e).__name__}: {str(e)[:100]}")
```

### Test 3.3: BART for Question Answering

```python
print("\n--- Testing BART (facebook/bart-base) ---")
try:
    bart_qa = pipeline('question-answering', model='facebook/bart-base', device=device)
    result_bart = bart_qa(question=question, context=context)
    print(f"Answer: {result_bart['answer']}")
    print(f"Score: {result_bart['score']:.4f}")
    print("Status: SUCCESS")
except Exception as e:
    print(f"Status: FAILURE")
    print(f"Error: {type(e).__name__}: {str(e)[:100]}")
```
